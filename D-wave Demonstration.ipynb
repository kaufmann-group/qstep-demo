## Review

Recall that Quantum Annealer's natively only solve a specific class of problems, namely Quadratic Binary Unconstrained Optimization (QUBO) problems. The hardware does so by implementing the Ising Hamiltonian via that has the form,

$$H(\mathbf{s}) = \sum_i h_i s_i + \sum_{i<j} J_{ij} s_i\, s_j, \quad s_i\in\{-1, +1\},$$

a simple linear transformation $\mathbb{s} \rightarrow \mathbb{x}$ allows us to write a QUBO objective function in terms of the Ising Hamiltonian,

$$Q(\mathbf{x}) = \sum_i a_i x_i + \sum_{i<j} b_{ij} x_i\, x_j, \quad x_i\in\{0, 1\}.$$

This powerful technique allows us to write a QUBO objective function in terms of a Ising Hamiltonian for D-wave systems to use an array of superconducting circuits to simulate the system and measure its ground state energy which corresponds to the minimum of the QUBO objective function.
