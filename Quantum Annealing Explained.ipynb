## Basic Quantum Mechanics

In classical computation bits can be either in a state of $0$ or $1$, however in quantum computing a **qubit** is a vector that takes in values on $\mathbb{C}^2$:

$$|\psi\rangle = \alpha|0\rangle + \beta|1\rangle, \qquad |0\rangle = \begin{pmatrix}1\\0\end{pmatrix}, \qquad |1\rangle = \begin{pmatrix}0\\1\end{pmatrix}, \qquad |\alpha|^2 + |\beta|^2 = 1$$

The **Born rule** states for the wavefunction $\ket{\psi}$ above, you get $0$ with probability $|\alpha|^2$ and $1$ with probability $|\beta|^2$, and measurement collapses the state into one of the two measurements.

The **Hamiltonain**, denoted $\mathcal{H}$ is a Hermetian matrix that encodes the energy of the system as its eigenvalues, while its eigenvectors are the states with those respective energies. The lowest eigenstate of the Hamiltonian is known as the **ground state**.
